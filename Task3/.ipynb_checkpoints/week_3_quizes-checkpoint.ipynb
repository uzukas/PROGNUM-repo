```python
masses = [1.9891e+30, 1.8986e+27, 
          5.6846e+26, 1.0243e+26, 8.6810e+25,
          5.9736e+24, 4.8685e+24, 6.4185e+23, 
          3.3022e+23, 7.349e+22, 1.25e+22]

new_masses = []

for i in masses[:]:
    if i >= masses[-2]:
        new_masses.append(i)

print(new_masses)

last_5_masses = masses[6:]

print(f"The average mass is {sum(last_5_masses)/len(last_5_masses)} kg")
```

```python
import math as m

G = 6.67e-11
v = []

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']

for i in zip(masses, radii):
    M = i[0]
    R = i[1]
    v.append(round(m.sqrt(2 * G * M / R), 2))
    print(f"Mass: {M:.2e}")

print(v)

planet_info = {}

for i in zip(names, v):
    planet_info[i[0]] = i[1]

print(planet_info)
```

```python
import math as m

def angular_size(D, d):
    delta = (2 * m.atan(d /(2 * D))) * (180/m.pi)
    return delta

print(angular_size(1, 60))
```

```python
import math as m

def cylinder(r,h):
    A = 2 * m.pi * r**2 + 2 * m.pi * r * h
    V = m.pi * r**2 * h
    return A, V

area, volume = cylinder(10, 30)
print(f"Area: {area}, Volume:{volume}, Ratio of Area to Volume: {area/volume}")
```
