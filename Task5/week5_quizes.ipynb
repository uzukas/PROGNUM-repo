```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("ngc6946.txt")

cube = data.reshape(40, 50, 70)

z_profile = cube[:,25,35]
print(z_profile)

z_max = np.max(z_profile)
z_min = np.min(z_profile)
z_sum = np.sum(z_profile)
z_mean = np.mean(z_profile)
z_std = np.std(z_profile)

plt.plot(z_profile, label="sd")
plt.text(0, z_max, f"Mean: {z_mean}, min: {z_min}, max: {z_max}, sum: {z_sum}")
plt.text(20, z_max - 0.005, f"std: {z_std}")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("ngc6946.txt")

cube = data.reshape(40, 50, 70)

first_plane = cube[0, :, :]

mean_val = np.mean(first_plane)
std_val = np.std(first_plane)

im = plt.imshow(first_plane, origin="lower", cmap="viridis")
plt.contour(first_plane, levels=[mean_val, mean_val + std_val, mean_val + 2*std_val], colors="white")
plt.colorbar(im)
```

```python
import scipy.constants as spc

class Galaxy:    
    """Class for redshift calculation of galaxies"""  
        
    caH_0 = 3968.5      # Calcium H line in rest frame
    c = spc.c * 0.001
    
    def __init__(self, name, calciumHline=caH_0):
        self.name = name
        self.caH = calciumHline
    def redshift(self):        
        z = (self.caH - self.caH_0)/self.caH
        return z
    def velocity(self):
        v = self.redshift() * self.c
        return v

g = Galaxy("Super Galaxy 102", calciumHline = 4000)

print(g.velocity())
```
